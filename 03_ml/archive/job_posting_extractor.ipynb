```python
import dspy
from pydantic import BaseModel
from typing import Union
from rich import print
```

```python
llm = dspy.OllamaLocal(model='mixtral:8x7b', max_tokens = 4000)
dspy.settings.configure(lm=llm)
```

```python
class JobPostingInput(BaseModel):
    jobpost: str

class JobExtraction(BaseModel):
    #education: Union[str, list[str]]
    certifications: Union[str, list[str]]
```

```python
class JobInfoExtractor(dspy.Signature):
    """Extract education and certifications within job posting"""
    jobpost: str = dspy.InputField(desc="may contain relevant facts related to education or certifications")
    info_extracted: JobExtraction = dspy.OutputField()

class PostingModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.job_extraction = dspy.TypedPredictor(JobInfoExtractor)
    
    def forward(self, jobpost):
        return self.job_extraction(jobpost=jobpost).info_extracted
```
