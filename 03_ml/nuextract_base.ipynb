## NuExtract Information Extraction

#### Load in Libraries

```python
import json
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch 
import gc
```

#### Create Prediction Function 
    * sample example from huggingface https://huggingface.co/numind/NuExtract
    * added two lines to clear memory to avoid OOM

```python
def predict_NuExtract(model, tokenizer, text, schema, example=["","",""]):
    schema = json.dumps(json.loads(schema), indent=4)
    input_llm =  "<|input|>\n### Template:\n" +  schema + "\n"
    for i in example:
      if i != "":
          input_llm += "### Example:\n"+ json.dumps(json.loads(i), indent=4)+"\n"
    
    input_llm +=  "### Text:\n"+text +"\n<|output|>\n"
    input_ids = tokenizer(input_llm, return_tensors="pt", truncation=True, max_length=5000).to("cuda")

    output = tokenizer.decode(model.generate(**input_ids)[0], skip_special_tokens=True)
    torch.cuda.empty_cache()
    gc.collect()
    return output.split("<|output|>")[1].split("<|end-output|>")[0]
```

#### Load in model 
    * blfoat16 takes 10 GB of memory 

```python
model = AutoModelForCausalLM.from_pretrained("numind/NuExtract", torch_dtype=torch.bfloat16, trust_remote_code=True).to("cuda")
tokenizer = AutoTokenizer.from_pretrained("numind/NuExtract", trust_remote_code=True)
```
