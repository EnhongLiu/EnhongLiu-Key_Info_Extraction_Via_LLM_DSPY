```python
import pandas as pd
import ast
from sentence_transformers import SentenceTransformer, util
from rich import print
```

```python
df = pd.read_csv("/Users/justinvhuang/Desktop/ISYE-CSE-MGT-6748-Group-1/data/sample_split_posting.csv", index_col = 0).reset_index(drop = True)
```

```python
text = ast.literal_eval(df.loc[1, 'sentence_split'])
```

```python
print(text)
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# Define example sentences for each category
skill_examples = ["skills required", "experience needed", "strong communication skills"]
location_examples = ["job is located", "work location", "based in"]
education_examples = ["degree required", "education qualification", "Master's degree"]
pay_examples = ["salary", "pay", "compensation"]

# Encode sentences and examples
sentence_embeddings = model.encode(text)
skill_embeddings = model.encode(skill_examples)
location_embeddings = model.encode(location_examples)
education_embeddings = model.encode(education_examples)
pay_embeddings = model.encode(pay_examples)
```

```python
# Function to find relevant sentences
def find_relevant_sentences(category_embeddings):
    relevant_sentences = []
    for sentence, embedding in zip(text, sentence_embeddings):
        similarity = util.pytorch_cos_sim(embedding, category_embeddings).mean().item()
        if similarity > 0.3:  
            relevant_sentences.append(sentence)
    return relevant_sentences

# Get relevant sentences for each category
skill_sentences = find_relevant_sentences(skill_embeddings)
location_sentences = find_relevant_sentences(location_embeddings)
education_sentences = find_relevant_sentences(education_embeddings)
pay_sentences = find_relevant_sentences(pay_embeddings)
```

```python
# Print results
print("Skill-related sentences:", skill_sentences)
print("Location-related sentences:", location_sentences)
print("Education-related sentences:", education_sentences)
print("Pay-related sentences:", pay_sentences)
```
