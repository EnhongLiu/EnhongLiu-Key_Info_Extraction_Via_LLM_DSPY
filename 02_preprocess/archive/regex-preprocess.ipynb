```python
import re
```

```python
# Extract education requirement
education_pattern = re.compile(r'\b(?:High school diploma or equivalent|degree|diploma)\b', re.IGNORECASE)
education = education_pattern.findall(text)
education = education[0] if education else "Not specified"

# Extract salary range
salary_pattern = re.compile(r'\$\d+\.?\d*K\s*-\s*\$\d+\.?\d*K', re.IGNORECASE)
salary = salary_pattern.findall(text)
salary = salary[0] if salary else "Not specified"

# Extract job skills
skills_pattern = re.compile(r'(?<=Requirements:)(.*?)(?=Benefits Offered:)', re.DOTALL | re.IGNORECASE)
skills = skills_pattern.findall(text)
skills = skills[0].strip().split('\n') if skills else ["Not specified"]
# Clean up skills list
skills = [skill.strip() for skill in skills if skill.strip()]
# Extract job location
location_pattern = re.compile(r'\b[A-Z][a-zA-Z\s]*, [A-Z]{2} \d{5}\b')
location = location_pattern.findall(text)
location = location[0] if location else "Not specified"

# Determine if job is remote, hybrid, or onsite
remote_pattern = re.compile(r'\b(remote|hybrid|onsite|on-site)\b', re.IGNORECASE)
work_type = remote_pattern.findall(text)
work_type = work_type[0] if work_type else "onsite"



# Display the extracted information
print("Education Requirement:", education)
print("Salary Range:", salary)
print("Job Skills:")
for skill in skills:
    print("-", skill)
print("Job Location:", location)
print("Work Type:", work_type)
```
