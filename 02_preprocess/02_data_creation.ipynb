## DataSetCreation - Create Custom Dataset for SetFit

#### Load in Python Libraries

```python
import os 
import pickle 
import pandas as pd
import mlflow
from tqdm import tqdm
```

#### Set Experiment

```python
mlflow.set_experiment("adeptID")
mlflow.start_run(run_name = "dataset-creation")
```

#### Helper Functions

```python
def load_pickles_from_folder(folder_path):

    loaded_objects = []

    for filename in tqdm(os.listdir(folder_path)):

        if filename.endswith(".pickle"):

            file_path = os.path.join(folder_path, filename)

            with open(file_path, 'rb') as file:

                loaded_objects.append(pickle.load(file))

    return loaded_objects

def create_dataframe(data):

    data_list = []
    
    for dictionary in tqdm(data):
        
        df_list = [(k, v) for k, vals in dictionary.items() for v in vals]

        df = pd.DataFrame(df_list, columns=['id', 'text'])

        df = df.reset_index().rename(columns={'index': 'order'})

        data_list.append(df)

    all_data = pd.concat(data_list, ignore_index=True)

    return all_data
```

#### Load in Pickle Files

```python
loaded_data = load_pickles_from_folder(r"C:\Users\jvhua\OneDrive\Desktop\ISYE-CSE-MGT-6748-Group-1\02_preprocess")
mlflow.log_param('number_pickle_files', len(loaded_data))
```

#### Create Pandas DataFrame

```python
df = create_dataframe(loaded_data)
mlflow.log_params({'rows_data_frame': df.shape[0], 'num_columns_data_frame': df.shape[1]})
```

#### Save Excel file for labeling

```python
sample_num = 10000
save_name = 'set-fit-labeling-data_v2.xlsx'
random_state = 7
df.sample(sample_num, random_state=7).to_excel(save_name)
mlflow.log_params({'num_sample': sample_num, 'save_path': save_name, 'random_state': random_state})
```

```python
mlflow.end_run()
```
