## SetFit Model Comparison

#### Load Python Libraries

```python
from setfit import SetFitModel
import pandas as pd
from tqdm import tqdm
import numpy as np
from rich import print
import mlflow 
from sklearn.metrics import accuracy_score
```

#### Set Experiment

```python
mlflow.set_experiment("adeptID")
mlflow.start_run(run_name = "setfit-model-distillation")
```

#### Load in SetFit Models

```python
model = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\models\setfit_mini_v3", trust_remote_code = True)
model2 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\models\setfit_300_nomic_model", trust_remote_code = True)
model3 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\models\setfit_model", trust_remote_code = True)
model4 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\models\setfit_nomic_model", trust_remote_code = True)
model5 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\ISYE-CSE-MGT-6748-Group-1\setfit_model_v5", trust_remote_code = True)
model6 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\ISYE-CSE-MGT-6748-Group-1\setfit_model_distilled_v3", trust_remote_code = True)
model7 = SetFitModel.from_pretrained(r"C:\Users\jvhua\OneDrive\Desktop\ISYE-CSE-MGT-6748-Group-1\sefit_model_v6", trust_remote_code = True)
```

#### Correct Answers for Custom Dataset

```python
custom_example_dataset = {'example1':"for this job you will need a AAA certification",
                          'example2':"this job has a salary of 100,000 a year",
                          'example3':"we are looking for team leadership",
                          'example4':"in this job it is a remote oppurtunity",
                          'example5':'you will be the best you available',
                          'example6':'we offer a salary of 20 dollars an hour',
                          'example7':'we do not consider age and gender or race we are an equal oppurtunity employer'}
y_true = [1, 1, 0, 1, 0, 1, 0]
```

#### Record Evaluation on Custom Dataset

```python
preds = model.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfitmini_v3', accuracy_score(preds,y_true))
preds = model2.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfit_300_nomic', accuracy_score(preds,y_true))
preds = model3.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfit_model', accuracy_score(preds,y_true))
preds = model4.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfit_nomic_model', accuracy_score(preds,y_true))
preds = model5.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfit_model_v5', accuracy_score(preds,y_true))
preds = model6.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('setfit_model_distilled_v3', accuracy_score(preds,y_true))
preds = model7.predict(list(custom_example_dataset.values()))
print(accuracy_score(preds,y_true))
mlflow.log_metric('sefit_model_v6', accuracy_score(preds,y_true))
```
